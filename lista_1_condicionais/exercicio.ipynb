## Exercicio 2 — Multa de radar (if + dois elif)

Um radar mede `velocidade_medida` numa via de `velocidade_maxima`.
Use **uma unica cadeia** com **dois `elif`** (a ordem importa!). Os
limites usam o excesso com percentual truncado (`int()` corta as
casas decimais) e ja vem calculados na funcao:

- `limite_leve`  = `int(1.10 * velocidade_maxima)` (10%)
- `limite_grave` = `int(1.50 * velocidade_maxima)` (50%)

| comparacao                                   | retorno        |
|----------------------------------------------|----------------|
| `velocidade_medida <= velocidade_maxima`     | `"ok"`         |
| acima do maximo, mas `<= limite_leve`        | `"leve"`       |
| acima de `limite_leve`, mas `<= limite_grave`| `"grave"`      |
| acima de `limite_grave`                      | `"gravissima"` |

Exemplos (limite 60): 60 -> `ok`; 61..66 -> `leve`; 67..90 -> `grave`;
91 ou mais -> `gravissima`.

## Exercicio 3 — Custo de hospedagem (faixas + desafio)

A rede `Sono Tranquilo` cobra `numero_diarias` vezes `preco_diaria`,
com desconto pela duracao da estadia:

**Parte 1 — desconto por duracao:**

| duracao                | fator                    |
|------------------------|--------------------------|
| ate 5 noites           | 1.0 (sem desconto)       |
| de 6 a 29 noites       | 0.90 (10% de desconto)   |
| de 30 noites ou mais   | 0.70 (30% de desconto)   |

**Parte 2 — desafio (+1 ponto):** se a estadia tem **exatamente 7
noites** **e** a reserva e numa **semana impar** do ano
(`numero_semana % 2 == 1`), multiplica o fator por `0.95` (mais 5%).

Regras gerais: retorne o custo total (numero); se `numero_diarias` ou
`preco_diaria` vierem zero ou negativos, retorne `0.0`; pequenas
diferencas de arredondamento sao aceitas pelo corretor.
